# Домашнее задание на ML метрики

## Обозначения

- $c$: Оптимальная константная модель.  
- $y_i$: Целевые значения в наборе данных.  
- $w_i$: Веса, связанные с функцией потерь.
- $q$: Значение квантиля в диапазоне $[0, 1]$.

## Задача 1

### Описание

Рассмотрим следующую функцию потерь:

\begin{align}
L = \sum w_i \cdot \left( \log(y_i) - \log(c) \right)^2
\end{align}

где:

- $\sum w_i = 1$

#### Задания

1. **Аналитически найдите лучшую константу $c$** для данной функции потерь.
2. **Определите название агрегации $y_i$**, которая получается в итоге, если все веса равны: $w_1 = w_2 = \dots = w_n$.

### Решение

#### 1. Аналитическое нахождение лучшей константы $c$

Чтобы найти оптимальное значение $c$, нам нужно минимизировать функцию потерь $L$ по отношению к $c$. Стандартный способ сделать это — взять производную $L$ по $c$ и приравнять её к нулю. Это позволит нам найти точку экстремума (в данном случае, минимума).

\begin{align}
\frac{\partial L}{\partial c} = \frac{\partial}{\partial c} \sum w_i \cdot \left( \log(y_i) - \log(c) \right)^2 = 0
\end{align}

Применим правило дифференцирования сложной функции (chain rule). Производная внешней функции $(...)^2$ равна $2 \cdot (...)$, а производная внутренней функции $(\log(y_i) - \log(c))$ по $c$ равна $(0 - \frac{1}{c})$.

\begin{align}
\frac{\partial L}{\partial c} = \sum w_i \cdot 2 \cdot \left( \log(y_i) - \log(c) \right) \cdot \left(-\frac{1}{c}\right)
\end{align}

Упростим выражение, вынеся общие множители за знак суммы:

\begin{align}
\frac{\partial L}{\partial c} = -\frac{2}{c} \sum w_i \cdot \left( \log(y_i) - \log(c) \right)
\end{align}

Теперь приравняем производную к нулю. Так как $c$ должно быть положительным (из-за логарифма), множитель $-2/c$ не равен нулю, поэтому мы можем приравнять к нулю только сумму:

\begin{align}
\sum w_i \cdot \left( \log(y_i) - \log(c) \right) = 0
\end{align}

Раскроем скобки и перегруппируем члены:

\begin{align}
\sum w_i \cdot \log(y_i) - \sum w_i \cdot \log(c) = 0 \\
\sum w_i \cdot \log(y_i) = \log(c) \cdot \sum w_i
\end{align}

По условию, сумма весов $\sum w_i = 1$. Подставим это в уравнение:

\begin{align}
\log(c) = \sum w_i \cdot \log(y_i)
\end{align}

Чтобы найти $c$, возьмем экспоненту от обеих частей уравнения:

\begin{align}
c = \exp\left(\sum w_i \cdot \log(y_i)\right)
\end{align}

Используя свойства логарифмов и экспонент ($a \cdot \log(b) = \log(b^a)$ и $\exp(a+b) = \exp(a) \cdot \exp(b)$), мы можем переписать это выражение. Таким образом, оптимальное $c$ является **взвешенным средним геометрическим** от $y_i$:

\begin{align}
c = \prod y_i^{w_i}
\end{align}

---

#### 2. Агрегация $y_i$, когда $w_1 = w_2 = \dots = w_n$

Если все веса равны, то для выполнения условия $\sum w_i = 1$ каждый вес должен быть равен $w_i = \frac{1}{n}$. В этом случае веса становятся равномерными, и формула для оптимального $c$ упрощается:

\begin{align}
c = \prod y_i^{1/n}
\end{align}

Это выражение является определением **среднего геометрического** для $y_i$.

---

### Итоговые ответы

1. Лучшая константа $c$ — это взвешенное среднее геометрическое:

   \begin{align}
   c = \prod y_i^{w_i}
   \end{align}

2. Когда $w_1 = w_2 = \dots = w_n$, агрегация $y_i$ называется **средним геометрическим**.

## Задача 2

### Описание

Рассмотрим **квантильную функцию потерь** $L$ и докажите, что оптимальная константа $c$ соответствует квантилю $q$ данных $y_1, \dots, y_n$.

Квантильная функция потерь определяется как:

\begin{align}
L =
\begin{cases}
q \cdot (y_i - c), & \text{если } y_i \geq c \\
(1-q) \cdot (c - y_i), & \text{если } y_i < c
\end{cases}
\end{align}

**Подсказка**:

Доказательство для оптимальной константы при MAE (средней абсолютной ошибке) можно найти на [этой странице](https://ds100.org/course-notes/constant_model_loss_transformations/loss_transformations.html). MAE является частным случаем квантильной потери при $q=0.5$.


### Решение

Наша цель — найти значение $c$, которое минимизирует общую квантильную потерю $L$ для всех точек данных.

1. Запишем общую функцию потерь $L(c)$ как сумму потерь по всем наблюдениям $y_i$. Мы разделим сумму на две части: для точек, где $y_i \geq c$, и для точек, где $y_i < c$.

   \begin{align}
   L(c) = \sum_{y_i \geq c} q \cdot (y_i - c) + \sum_{y_i < c} (1-q) \cdot (c - y_i)
   \end{align}

2. Продифференцируем $L(c)$ по $c$ и найдем значение, при котором производная равна нулю. Хотя эта функция не является гладкой (она имеет "изломы" в точках $y_i = c$), мы можем рассмотреть её производную по частям для всех $c$, не равных какому-либо $y_i$.

   - Для первого слагаемого $\sum_{y_i \geq c} q \cdot (y_i - c)$, производная по $c$ для каждого члена суммы равна $-q$. Таким образом, производная всей суммы:

     \begin{align}
     \frac{\partial}{\partial c} \sum_{y_i \geq c} q \cdot (y_i - c) = \sum_{y_i \geq c} (-q) = -q \cdot N_{\geq c}
     \end{align}

     где $N_{\geq c}$ — это количество элементов в множестве $\{y_i \mid y_i \geq c\}$.

   - Для второго слагаемого $\sum_{y_i < c} (1-q) \cdot (c - y_i)$, производная по $c$ для каждого члена суммы равна $(1-q)$. Таким образом, производная всей суммы:

     \begin{align}
     \frac{\partial}{\partial c} \sum_{y_i < c} (1-q) \cdot (c - y_i) = \sum_{y_i < c} (1-q) = (1-q) \cdot N_{< c}
     \end{align}

     где $N_{< c}$ — это количество элементов в множестве $\{y_i \mid y_i < c\}$.

3. Полная производная $L(c)$ является суммой этих двух частей:

   \begin{align}
   \frac{\partial L}{\partial c} = -q \cdot N_{\geq c} + (1-q) \cdot N_{< c}
   \end{align}

4. Чтобы минимизировать потери, приравняем производную к нулю:

   \begin{align}
   -q \cdot N_{\geq c} + (1-q) \cdot N_{< c} = 0
   \end{align}

   Перегруппируем члены:

   \begin{align}
   (1-q) \cdot N_{< c} = q \cdot N_{\geq c}
   \end{align}

5. Выразим $N_{\geq c}$ через общее число точек $N = N_{< c} + N_{\geq c}$:

   \begin{align}
   (1-q) \cdot N_{< c} &= q \cdot (N - N_{< c}) \\
   N_{< c} - q \cdot N_{< c} &= q \cdot N - q \cdot N_{< c} \\
   N_{< c} &= q \cdot N \\
   \frac{N_{< c}}{N} &= q
   \end{align}

Это уравнение означает, что значение $c$, которое минимизирует $L$, делит данные таким образом, что доля точек, которые меньше $c$, равна $q$. По определению, это и есть **квантиль уровня $q$** данных.

---

### Итоговый ответ

Оптимальная константа $c$ для квантильной функции потерь это:

\begin{align}
c = \text{quantile}(y, q)
\end{align}

## Задача 3: Найти оптимальную константу методом перебора (Brute Force)

### Пояснение
В этой задаче мы напишем функцию, которая находит оптимальную константу `c` для заданной функции потерь, используя метод полного перебора (brute force). Этот метод заключается в простом переборе множества возможных значений для `c` в заданном диапазоне (от `min(y)` до `max(y)`) с небольшим шагом `tol` и выборе того значения, которое дает наименьшие потери.

В качестве примера используется функция потерь $L(y, c) = \sum \log(\cosh(y_i - c))$. Эта функция, известная как **Log-Cosh**, является гладкой аппроксимацией MAE (Mean Absolute Error). Она ведет себя как MSE для малых ошибок и как MAE для больших, что делает ее менее чувствительной к выбросам по сравнению с MSE.

```python
import numpy as np

def bruteforce_constant(y: np.ndarray, loss_func: callable, tol: float) -> float:
    """
    Находит оптимальную константу c методом полного перебора, используя заданную функцию потерь.

    Параметры:
    y (np.ndarray): Массив целевых значений.
    loss_func (callable): Функция, которая вычисляет потери для заданных y и c.
    tol (float): Размер шага для генерации потенциальных значений c между min(y) и max(y).

    Возвращает:
    float: Оптимальная константа c, минимизирующая потери.
    """
    # Генерируем потенциальные значения c с помощью np.arange
    c_values = np.arange(min(y), max(y), tol)

    # Инициализируем переменные для отслеживания минимальных потерь и соответствующего c
    min_loss = float('inf')
    best_c = None

    # Вычисляем потери для каждого c и находим минимум
    for c in c_values:
        current_loss = loss_func(y, c)
        if current_loss < min_loss:
            min_loss = current_loss
            best_c = c

    return best_c

# Определяем функцию потерь
def loss(y, c):
    return np.sum(np.log(np.cosh(y - c)))

# Пример использования функции
y = np.array([1, 2, 3, 4, 55, 99, 100])
optimal_c = bruteforce_constant(y, loss, tol=0.001)
min_loss = loss(y, optimal_c)
print(f"Оптимальная константа c: {optimal_c}")
print(f"Минимальные потери: {min_loss}")
```

## Задача 4: Найти оптимальную константу с помощью оптимизации

### Пояснение
Вместо полного перебора, который может быть медленным и неточным, мы будем использовать более эффективный метод — численную оптимизацию с помощью функции `scipy.optimize.minimize`. Этот подход использует градиентные или другие численные методы для быстрого и точного нахождения минимума функции. Мы передаем ей нашу функцию потерь, начальное предположение для `c` и некоторые параметры, такие как точность (`tol`).

```python
from scipy.optimize import minimize
import numpy as np

def minimize_constant(y: np.ndarray, loss_func: callable, tol: float) -> float:
    """
    Находит оптимальную константу c, используя функцию minimize из scipy.

    Параметры:
    y (np.ndarray): Массив целевых значений.
    loss_func (callable): Функция, которая вычисляет потери для заданных y и c.
    tol (float): Точность для процесса оптимизации.

    Возвращает:
    float: Оптимальная константа c, минимизирующая потери.
    """
    # Начальное предположение для c - среднее значение y
    initial_guess = np.mean(y)

    # Определяем целевую функцию для минимизации
    def objective(c):
        return loss_func(y, c)

    # Используем функцию minimize из scipy для поиска оптимального c
    result = minimize(
        objective,          # функция, которую минимизируем
        initial_guess,      # начальная точка
        tol=tol,            # требуемая точность
        bounds=[(min(y), max(y))], # границы поиска для c
        method='trust-constr' # метод оптимизации
    )

    return result.x[0]

# Определяем функцию потерь
def loss(y, c):
    return np.sum(np.log(np.cosh(y - c)))

# Пример использования функции
y = np.array([1, 2, 3, 4, 55, 99, 100])
optimal_c = minimize_constant(y, loss, tol=0.001)
min_loss = loss(y, optimal_c)
print(f"Оптимальная константа c: {optimal_c}")
print(f"Минимальные потери: {min_loss}")
```

# Задача 5

## Описание
В задаче многоклассовой классификации мы часто сравниваем производительность модели с простыми базовыми подходами (baseline). Два таких подхода:

1. **Всегда предсказывать самый частый класс** (также известный как “константный baseline”).  
2. **Случайно выбирать класс** для каждого объекта с вероятностями, пропорциональными частотам классов в обучающем наборе.

Цель этой задачи — определить, какой из этих двух подходов дает более высокую **точность (Accuracy)** на данном наборе данных.

Вам дано:  
1. Набор объектов $( X )$ и их истинные метки классов $( y )$, где $( y \in \{C_1, C_2, \dots, C_k\} )$.  
2. Частоты (или доли) каждого класса в обучающих данных.

Вам нужно:  
1. **Объяснить**, как реализовать эти два базовых подхода (всегда предсказывать самый частый класс против случайного выбора класса в соответствии с его частотой).  
2. **Рассчитать** ожидаемую точность (Accuracy) для каждого подхода.  
3. **Сравнить** результаты, чтобы определить, какой метод с большей вероятностью даст более высокую точность.  
4. **Обсудить**, почему один подход может быть более выгодным, чем другой, с точки зрения точности.

## Решение

### Шаг 1. Краткое объяснение
Мы сравниваем два базовых подхода: один, который всегда предсказывает самый частый класс, и другой, который предсказывает классы случайным образом в соответствии с их частотами в данных.

### Шаг 2. Описание базовых подходов (Baselines)
- **Константный Baseline (самый частый класс)**: Определяем класс $( C_{\max} )$, который встречается чаще всего в обучающем наборе. Затем для любого входного объекта $( x_i )$ мы всегда предсказываем $( C_{\max} )$.  
- **Случайный Baseline (Random Sampling)**: Для каждого входного объекта $( x_i )$, мы случайно выбираем класс $( C_j )$ с вероятностью, равной его доле в данных $( p_j = \frac{n_j}{N})$, где $( n_j )$ — количество примеров класса $( C_j )$ в обучающем наборе, а $( N )$ — общее количество примеров.

### Шаг 3. Расчет точности (Accuracy)
- **Точность (самый частый класс)**  
  Точность этого подхода равна доле самого частого класса, так как модель будет права только для объектов этого класса.
  $$
  \mathrm{Accuracy}_{\mathrm{const}} = p_{\max},
  $$
  где $( p_{\max} )$ — это доля самого частого класса.

- **Точность (случайный выбор)**  
  Ожидаемая точность — это вероятность случайного выбора правильной метки. Для любого объекта, истинный класс которого $( C_j )$, вероятность угадать его равна $( p_j )$. Так как доля таких объектов в данных равна $( p_j )$, то общая вероятность угадать правильный класс для случайно выбранного объекта равна сумме произведений этих вероятностей по всем классам.
  $$
  \mathrm{Accuracy}_{\mathrm{random}} = \sum_{j=1}^{k} p_j \cdot p_j = \sum_{j=1}^{k} p_j^2.
  $$

### Шаг 4. Математическое сравнение
Пусть $( p_1, p_2, \dots, p_k )$ — доли классов, такие что  
$$
p_1 + p_2 + \cdots + p_k = 1,
$$  
и пусть  
$$
p_{\max} = \max_{1 \le j \le k} \, p_j.
$$  

Тогда  
$$
\mathrm{Accuracy}_{\mathrm{const}} = p_{\max},
\quad
\mathrm{Accuracy}_{\mathrm{random}} = \sum_{j=1}^{k} p_j^2.
$$  

Мы хотим проверить, верно ли неравенство:  
$$
p_{\max} \;\overset{?}{\ge}\; \sum_{j=1}^{k} p_j^2.
$$

**Доказательство**  
1. Заметим, что для любого класса $j$, его доля $p_j$ не превышает максимальную долю $p_{\max}$: $( p_j \le p_{\max} )$.  
2. Рассмотрим сумму квадратов долей:

   $$
   \sum_{j=1}^{k} p_j^2
   $$  
   Поскольку $p_j \le p_{\max}$, то $p_j^2 \le p_j \cdot p_{\max}$. Применим это неравенство к каждому члену суммы:
   $$
   \sum_{j=1}^{k} p_j^2 \;\le\; \sum_{j=1}^{k} p_j \cdot p_{\max}
   $$  
   Вынесем $p_{\max}$ за знак суммы, так как это константа:
   $$
   \sum_{j=1}^{k} p_j^2 \;\le\; p_{\max} \cdot \sum_{j=1}^{k} p_j
   $$  
   Так как сумма всех долей равна 1 ($\sum p_j = 1$), получаем:
   $$
   \sum_{j=1}^{k} p_j^2 \;\le\; p_{\max} \cdot 1 = p_{\max}
   $$  
3. Таким образом, мы доказали, что
   $$
   \sum_{j=1}^{k} p_j^2 \;\le\; p_{\max},
   $$
   что означает, что точность константного baseline (предсказание самого частого класса) всегда не ниже, чем у случайного baseline.

4. **Когда достигается равенство?** Равенство возможно в двух случаях:
   - Когда $( p_{\max} = 1 )$ (вырожденный случай, когда в данных есть только один класс). Тогда $\sum p_j^2 = 1^2 = 1$, и $p_{\max} = 1$.
   - Когда все доли $( p_j )$ равны, то есть $( p_j = \frac{1}{k} )$ для каждого $( j )$. В этом случае $( p_{\max} = \frac{1}{k} )$, а $( \sum_{j=1}^{k} p_j^2 = k \cdot \left(\frac{1}{k^2}\right) = \frac{1}{k} )$. Равенство выполняется.

### Итоговый ответ
В общем случае, предсказание самого частого класса **всегда дает точность, которая как минимум не ниже**, чем у случайного выбора класса с вероятностью, равной его доле. Равенство достигается только в крайних случаях, описанных выше. Следовательно, с точки зрения ожидаемой точности (Accuracy), константный baseline (предсказание $( C_{\max} )$) является более предпочтительным.
